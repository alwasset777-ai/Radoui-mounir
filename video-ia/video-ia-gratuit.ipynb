### 📌 الرابط الثابت ديالك (حطو فالمفضلة فالتليفون)
**https://ntfy.sh/wasset-video-1b0e4427b990f558**
فيه ديما آخر رابط ديال التطبيق + إشعار ملي يوجد الفيديو.

### مرة وحدة فقط
1. وثّق الحساب ديال Kaggle برقم الهاتف (بلا هادشي ما كاينش GPU ولا أنترنيت).
2. فهاد الصفحة (Settings): **Accelerator ← GPU T4 x2** و **Internet ← On**.

### كل مرة بغيتي تخدم
- كليكي **Save Version ← Save & Run All (Commit) ← Save**، وسد الصفحة عادي (كيخدم فالخلفية حتى 12 ساعة).
- حل الرابط الثابت فوق، وتسنى ميساج «🔗 التطبيق خدام» (2-4 دقايق)، وكليكي عليه.
- التطبيق كيطفى بوحدو إلا ما استعملتيهش 60 دقيقة (باش توفر الساعات)، والفيديوهات كيتحفظو فـ **Output** ديال هاد الصفحة.

⏱️ Kaggle كيعطي حوالي 30 ساعة GPU فالسيمانة فابور. أول تحميل للموديل 5-15 دقيقة.


```python
# ▶️ هادي هي الخلية الوحيدة. ما تبدل فيها والو.
NTFY_TOPIC = "wasset-video-1b0e4427b990f558"   # الرابط الثابت: https://ntfy.sh/wasset-video-1b0e4427b990f558
IDLE_MINUTES = 60         # كيطفى بوحدو من بعد 60 دقيقة بلا استعمال

import os, sys, shutil, subprocess, base64, zlib, json, urllib.request
from collections import deque

def phone(title, msg):
    """إشعار فالتليفون (صفحة ntfy) باش تعرف شنو وقع حتى إلا كانت Kaggle مسدودة."""
    try:
        body = json.dumps({"topic": NTFY_TOPIC, "title": title, "message": msg[-3000:], "priority": 4,
                           "tags": ["warning"]}).encode()
        req = urllib.request.Request(os.environ.get("NTFY_SERVER", "https://ntfy.sh/"), data=body,
                                     headers={"Content-Type": "application/json"})
        urllib.request.urlopen(req, timeout=15)
    except Exception as e:
        print("notify failed:", e)
print("Python", sys.version.split()[0], flush=True)
gpu = shutil.which("nvidia-smi") and subprocess.run(
    ["nvidia-smi", "--query-gpu=name,memory.total", "--format=csv,noheader"], capture_output=True, text=True).stdout.strip()
if not gpu:
    phone("❌ ما كاينش GPU", "فـ Kaggle: Settings ← Accelerator ← GPU T4 x2، ومن بعد Save & Run All.")
    raise SystemExit("❌ ما كاينش GPU. Settings ← Accelerator ← GPU T4 x2، وعاود.")
print("✅ GPU:", gpu.replace("\n", " | "), flush=True)

print("⏳ كنثبت الأدوات (2-4 دقايق)…", flush=True)
r = subprocess.run([sys.executable, "-m", "pip", "install", "-q", "diffusers==0.42.1", "transformers==5.22.2",
                    "gradio==6.31.0", "accelerate", "deep-translator", "ftfy", "sentencepiece", "imageio",
                    "imageio-ffmpeg", "requests"], capture_output=True, text=True)
if r.returncode != 0:
    print(r.stdout[-2000:], r.stderr[-2000:])
    phone("❌ التثبيت ما نجحش", "تأكد بلي Internet ← On فـ Kaggle.\n" + r.stderr[-1500:])
    raise SystemExit("❌ التثبيت ما نجحش. تأكد بلي Internet ← On.")
print("✅ الأدوات واجدين", flush=True)

# كنجيبو آخر نسخة ديال التطبيق من GitHub (فيها آخر الإصلاحات)، وإلا ما خدماتش كنخدمو بالنسخة اللي هنا
APP_URL = "https://raw.githubusercontent.com/alwasset777-ai/Radoui-mounir/video-ia/video-ia/app.py"
APP = ""
try:
    import urllib.request
    code = urllib.request.urlopen(APP_URL, timeout=30).read()
    compile(code, "app.py", "exec")
    open("app.py", "wb").write(code)
    print("✅ آخر نسخة ديال التطبيق", flush=True)
except Exception as e:
    open("app.py", "wb").write(zlib.decompress(base64.b64decode(APP)))
    print("ℹ️ نسخة احتياطية ديال التطبيق:", e, flush=True)

env = dict(os.environ, NTFY_TOPIC=NTFY_TOPIC, IDLE_MINUTES=str(IDLE_MINUTES), PYTHONUNBUFFERED="1")
p = subprocess.Popen([sys.executable, "-u", "app.py"], env=env, stdout=subprocess.PIPE,
                     stderr=subprocess.STDOUT, text=True)
tail = deque(maxlen=25)
for line in p.stdout:
    print(line, end="", flush=True)
    tail.append(line)
p.wait()
if p.returncode != 0:  # وقف بسبب مشكل (ماشي الإطفاء العادي)
    phone(f"❌ التطبيق وقف بسبب مشكل (code {p.returncode})", "صيفط تصويرة ديال هاد الميساج لـ Claude:\n" + "".join(tail))
print("🔌 التطبيق طفا. الفيديوهات فـ Output (videos).")
```
